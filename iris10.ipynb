```python
numeric_features = [
    'SepalLengthCm',
    'SepalWidthCm',
    'PetalLengthCm',
    'PetalWidthCm'
]
```

```python
nominal_features = ['Species']

print("\nNumeric Features:")
print(numeric_features)

print("\nNominal Features:")
print(nominal_features)
```

```python
#histogram
```

```python
for column in numeric_features:

    plt.figure(figsize=(6,4))

    plt.hist(df[column],
             bins=10,
             color='skyblue',
             edgecolor='black')

    plt.title(f'Histogram of {column}')
    plt.xlabel(column)
    plt.ylabel('Frequency')

    plt.show()

# ---------------------------------------------------
# 3. Boxplots for Each Feature
# ---------------------------------------------------

plt.figure(figsize=(10,6))

df.boxplot(column=numeric_features)

plt.title('Boxplot of Iris Features')
plt.ylabel('Values (cm)')

plt.show()

# ---------------------------------------------------
# Outlier Detection using IQR Method
# ---------------------------------------------------

def detect_outliers(data, column):

    # Calculate Quartiles
    Q1 = data[column].quantile(0.25)
    Q3 = data[column].quantile(0.75)

    # Calculate IQR
    IQR = Q3 - Q1

    # Define Limits
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    # Find Outliers
    outliers = data[
        (data[column] < lower_bound) |
        (data[column] > upper_bound)
    ]

    print(f"\nFeature: {column}")
    print("Lower Bound:", lower_bound)
    print("Upper Bound:", upper_bound)

    print("\nOutliers:")
    print(outliers[[column, 'Species']])

    print("\nOutlier Row Indexes:")
    print(outliers.index.tolist())

# Detect Outliers for all Numeric Features
for column in numeric_features:
    detect_outliers(df, column)

# ---------------------------------------------------
# Species-wise Boxplot (Extra Analysis)
# ---------------------------------------------------

plt.figure(figsize=(8,5))

sns.boxplot(
    x='Species',
    y='PetalWidthCm',
    data=df,
    palette='Set2'
)

plt.title('Species-wise Boxplot of PetalWidthCm')

plt.show()

# ---------------------------------------------------
# Observations / Inference
# ---------------------------------------------------

print("\nObservations:")

print("1. All flower measurements are numeric features.")
print("2. Species is a nominal/categorical feature.")
print("3. Petal Length and Petal Width show greater variation among species.")
print("4. Iris-setosa flowers generally have smaller petals.")
print("5. Boxplots help in identifying outliers clearly.")
print("6. Some outliers are present in SepalWidthCm.")
print("7. Histograms show that most features are approximately normally distributed.")
```
