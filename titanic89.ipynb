```python
# Data Visualization I & II 



import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = sns.load_dataset("titanic")
df
```

```python
df.info()
```

```python
numeric_df = df.select_dtypes(include=["number"])
corr_matrix = numeric_df.corr()
corr_matrix
```

```python
sns.countplot(x='pclass', hue='survived', data=df)
plt.title('Survival Count by Passenger Class')
plt.xlabel('Passenger Class')
plt.ylabel('Survival Count')
plt.show()
```

```python
sns.barplot(x="sex", y="survived", data=df)
plt.title("Survival Rate by Gender")
plt.xlabel("Gender")
plt.ylabel("Survival Rate")
plt.show()
```

```python
sns.histplot(x='age', hue='survived', data=df, multiple="stack")
plt.title('Survival Count by Age')
plt.xlabel('Age')
plt.ylabel('Survival Count')
plt.show()
```

```python
sns.barplot(x="embarked", y="survived", data=df, errorbar=None)
plt.title("Survival Rate by Embarkation Port")
plt.xlabel("Port of Embarkation")
plt.ylabel("Survival Rate")
plt.show()
```

```python
sns.boxplot(x="survived", y="fare", data=df)
plt.title("Fare Distribution for Survivors and Non-Survivors")
plt.xlabel("Survival Status")
plt.ylabel("Fare Price")
plt.show()
```

```python
sns.barplot(x="sibsp", y="survived", data=df, errorbar=None)
plt.title("Survival Rate by Siblings/Spouses")
plt.xlabel("Siblings/Spouses")
plt.ylabel("Survival Rate")
plt.show()
```

```python
order = df["deck"].value_counts().index
sns.barplot(x="deck", y="survived", data=df, order=order, errorbar=None)
plt.title("Survival Rate by Deck")
plt.xlabel("Deck")
plt.ylabel("Survival Rate")
plt.show()
```

```python
sns.barplot(x="embark_town", y="survived", data=df, errorbar=None)
plt.title("Survival Rate by Embarkation Town")
plt.xlabel("Embark Town")
plt.ylabel("Survival Rate")
plt.show()
```
